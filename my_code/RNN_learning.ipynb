```python
import torch
import torch.nn as nn

if torch.cuda.is_available():
    device = torch.device("cuda:0") 
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print(f"Connect to device: {device}")
```

```python
batch_size = 4
seq_len = 10
embed_dim = 32
hidden_dim = 64

x = torch.randn(batch_size, seq_len, embed_dim) # shape = (B, L, D_in)

rnn = nn.RNN(
    input_size=embed_dim, # in: 32
    hidden_size=hidden_dim, # out: 64
    num_layers=1,
    batch_first=True,
)

output, h_n = rnn(x)

print(output.shape) # shape = (B, L, D_out)
print(h_n.shape) # first layer shape = (hidden layer#, B, D_out)
```

```python
# c_n: long short term memory
lstm = nn.LSTM(
    input_size=embed_dim,
    hidden_size=hidden_dim,
    num_layers=1,
    batch_first=True,
)
# x = (B, L, D_in)
output, (h_n, c_n) = lstm(x)

print(output.shape) # (B, L, D_out)
print(h_n.shape) # (hidden layer#, B, D_out)
print(c_n.shape) # (hidden layer#, B, D_out)
```
